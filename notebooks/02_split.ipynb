```python
import pandas as pd
from sklearn.model_selection import train_test_split

df = pd.read_csv("../donnees/creditcard.csv")
print(df.shape)
```

```python
X = df.drop(columns="Class")
y = df["Class"]

print("X :", X.shape)
print("y :", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=42,
)
```

```python
print("Train :", X_train.shape, "| Test :", X_test.shape)
print(f"Fraudes train : {y_train.sum()} ({y_train.mean():.3%})")
print(f"Fraudes test  : {y_test.sum()} ({y_test.mean():.3%})")
```

```python
import os
os.makedirs("../donnees/processed", exist_ok=True)

train = X_train.copy()
train["Class"] = y_train
train.to_csv("../donnees/processed/train.csv", index=False)

test = X_test.copy()
test["Class"] = y_test
test.to_csv("../donnees/processed/test.csv", index=False)

print("Fichiers sauvegardés")
```
